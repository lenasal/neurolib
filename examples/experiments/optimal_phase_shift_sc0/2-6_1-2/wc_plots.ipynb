```python
import matplotlib.pyplot as plt
import numpy as np
import os, scipy, pickle
import matplotlib

if os.getcwd().split(os.sep)[-1] == "examples":
    os.chdir('..')

# We import the model, stimuli, and the optimal control package
from neurolib.models.wc import WCModel
from neurolib.utils.stimulus import ZeroInput
from neurolib.optimal_control import oc_wc
from neurolib.optimal_control import cost_functions as cost_functions
from neurolib.optimal_control.oc_utils.plot_oc import ops_plotall, ops_plotsubset, ops_plot_acp_inits, ops_plot_timing_inits, ops_plot_timing_inits_L2, get_prc, plot_pcr
import scipy

# This will reload all imports as soon as the code changes
%load_ext autoreload
%autoreload 2

# We define colors for later convenience
color_x = 'red'
color_y = 'blue'
fs_=8
plt.rcParams["figure.figsize"] = (10,6)
plt.rcParams["axes.titlesize"] = fs_
plt.rcParams["axes.labelsize"] = fs_
matplotlib.rc('xtick', labelsize=fs_) 
matplotlib.rc('ytick', labelsize=fs_) 
plt.rc('legend',fontsize=fs_)
```

```python
wc = WCModel()
wc.params.sigma_ou = 0.

dt = 0.1
wc.params.dt = dt

duration = 300.
zero_input = ZeroInput().generate_input(duration=duration+dt, dt=dt)

wc.params["duration"] = duration
x_init = 0.4543040513180053 
y_init = 0.3987237035596491
wc.params["exc_init"] = np.array([[x_init]])
wc.params["inh_init"] = np.array([[y_init]])

wc.params["exc_ext"] = zero_input + 2.6
wc.params["inh_ext"] = zero_input + 1.2

wc.run()
peaks = scipy.signal.find_peaks(wc.exc[0,:])[0]
p_list = []
for i in range(3,len(peaks)):
    p_list.append(peaks[i]- peaks[i-1])
period = np.mean(p_list)*dt

savepath = ('..' + os.sep + '..' + os.sep + '..' + os.sep + '..' + os.sep + '..' + os.sep + 'writing' + os.sep +
                'paper2_chaos' + os.sep + 'exploration' + os.sep + 'images_ops' + os.sep + 'sc0' + os.sep + '2-6_1-2' + os.sep)

savepath_subset = ( '..' + os.sep + '..' + os.sep + '..' + os.sep + '..' + os.sep + '..' + os.sep + 'writing' + os.sep +
                        'paper2_chaos' + os.sep + 'opt_phase_shift' + os.sep + 'images' + os.sep + 'sc0' + os.sep + '2-6_1-2' + os.sep)
```

```python
wc = WCModel()
dist_array = [100.]

ep, em, ip, im, epim, emip, period = get_prc(wc, [2.6, 1.2], dist_array)
```

```python
dt = 0.01

for i in range(len(dist_array)):
    min0, min1 = np.amin(np.abs(ep[i,:750])), np.amin(np.abs(ep[i,750:]))
    i0 = np.where(min0 == ep[i,:750])
    i1 = np.where(min1 == ep[i,750:])

    i00 = i0[0][0] *2/(period/dt)
    i01 = i0[0][-1] *2/(period/dt)
    i10 = (i1[0][0]+750) *2/(period/dt)
    i11 = (i1[0][-1]+750) *2/(period/dt)

    print("E+")

    print(f"advance : %5.4f, %5.4f" % ( (i10 + i01)/2, i10-i01) )
    print(f"delay : %5.4f, %5.4f"% ((i00 +2+ i11)/2, i00+2.-i11))

    min0, min1 = np.amin(np.abs(em[i,:750])), np.amin(np.abs(em[i,750:]))
    i0 = np.where(min0 == em[i,:750])
    i1 = np.where(min1 == em[i,750:])

    i00 = i0[0][0] *2/(period/dt)
    i01 = i0[0][-1] *2/(period/dt)
    i10 = (i1[0][0]+750) *2/(period/dt)
    i11 = (i1[0][-1]+750) *2/(period/dt)

    print("E-")

    print(f"advance : %5.4f, %5.4f" % ( (i10 + i01)/2, i10-i01) )
    print(f"delay : %5.4f, %5.4f" % ((i00 +2+ i11)/2, i00+2.-i11))
    print("----------------------------------------")

    min0, min1 = np.amin(np.abs(ip[i,:750])), np.amin(np.abs(ip[i,750:]))
    i0 = np.where(min0 == ip[i,:750])
    i1 = np.where(min1 == ip[i,750:])

    i00 = i0[0][0] *2/(period/dt)
    i01 = i0[0][-1] *2/(period/dt)
    i10 = (i1[0][0]+750) *2/(period/dt)
    i11 = (i1[0][-1]+750) *2/(period/dt)

    print("I+")

    print(f"advance : %5.4f, %5.4f" % ( (i10 + i01)/2, i10-i01) )
    print(f"delay : %5.4f, %5.4f"% ((i00 +2+ i11)/2, i00+2.-i11))

    min0, min1 = np.amin(np.abs(im[i,:750])), np.amin(np.abs(im[i,750:]))
    i0 = np.where(min0 == im[i,:750])
    i1 = np.where(min1 == im[i,750:])

    i00 = i0[0][0] *2/(period/dt)
    i01 = i0[0][-1] *2/(period/dt)
    i10 = (i1[0][0]+750) *2/(period/dt)
    i11 = (i1[0][-1]+750) *2/(period/dt)

    print("I-")

    print(f"advance : %5.4f, %5.4f" % ( (i10 + i01)/2, i10-i01) )
    print(f"delay : %5.4f, %5.4f" % ((i00 +2+ i11)/2, i00+2.-i11))
    print("----------------------------------------")
```
